```python
from evds import evdsAPI
import pandas as pd
from datetime import datetime, timedelta

# 🔑 EVDS API key (https://evds2.tcmb.gov.tr/ adresinden ücretsiz al)
API_KEY = "BURAYA_API_KEYİNİ_YAZ"

# API bağlantısı
evds = evdsAPI(API_KEY)

# 📅 Bugünün tarihi
end_date = datetime.today()
start_date = end_date - timedelta(days=365)

# Tarih formatı (gün-ay-yıl)
start_date_str = start_date.strftime("%d-%m-%Y")
end_date_str = end_date.strftime("%d-%m-%Y")

# 📊 TÜFE verisi (aylık)
data = evds.get_data(
    data_series=['TP.FE.OKT'],  # TÜFE (2003=100)
    startdate=start_date_str,
    enddate=end_date_str,
    frequency=5  # aylık veri
)

# Kolon ismini düzenle
data.rename(columns={"TP_FE_OKT": "TUFE"}, inplace=True)

# Tarih sütununu datetime yap
data['Tarih'] = pd.to_datetime(data['Tarih'])

# Son 12 ayı kesin almak için sıralayıp son 12 satırı al
data = data.sort_values("Tarih").tail(12)

# 📈 Aylık enflasyon (%) hesapla
data["Aylik_Enflasyon_%"] = data["TUFE"].pct_change() * 100

# 📈 Yıllık enflasyon (%) hesapla
data["Yillik_Enflasyon_%"] = data["TUFE"].pct_change(12) * 100

# Excel olarak kaydet
file_name = "turkiye_enflasyon_son_12_ay.xlsx"
data.to_excel(file_name, index=False)

print("✅ Veri başarıyla çekildi ve kaydedildi:", file_name)
print(data)
```
